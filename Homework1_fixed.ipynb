## 4. Strings: Count Word Occurrences

**Challenge** (uses a function, which you will learn next session):

Count how many times the word "wood" appears in the text. The `clean()` function removes everything that is not a letter and replaces it with a space.

```python
text = """How much wood would a woodchuck chuck
If a woodchuck could chuck wood?
He would chuck, he would, as much as he could,
And chuck as much as a woodchuck would
If a woodchuck could chuck wood."""

def clean(s):
    news = ""
    s = s.lower()
    for c in s:
        if c >= "a" and c <= "z":
            news += c
        else:
            news += " "
    return news

count = 0
for word in clean(text).split():
    if word == "wood":
        count += 1

print(f'Number of times "wood" occurs: {count}')
```

## 6. Lists: Build and Shuffle a Card Deck

Build a list of 52 cards, then shuffle it using the Fisher-Yates algorithm.

**Exercise:** Run the cell. Then try: print only the Hearts.

```python
from random import randint

deck = []
for value in ("Ace", "2", "3", "4", "5", "6", "7", "8", "9", "10", "Jack", "Queen", "King"):
    for suit in ("Hearts", "Spades", "Clubs", "Diamonds"):
        deck.append(value + " of " + suit)

print(f"Deck has {len(deck)} cards")
print("First 5:", deck[:5])

# Shuffle (Fisher-Yates)
for i in range(len(deck)):
    j = randint(i, len(deck) - 1)
    deck[i], deck[j] = deck[j], deck[i]

print("After shuffle, first 5:", deck[:5])
```

## 8. Lists: Letter Frequency

Count how often each letter appears in a text, then sort by frequency (most common first).

**Challenge:** The `sort(key=...)` part uses a function, which you will learn next session. Try to follow the logic.

```python
text = "Now, it's quite simple to defend yourself against a man armed with a banana. First of all you force him to drop the banana; then, second, you eat the banana, thus disarming him. You have now rendered him helpless."

def count_letter(x):
    return x[0], -ord(x[1])

countlist = []
for i in range(26):
    countlist.append([0, chr(ord("a") + i)])

for letter in text.lower():
    if letter >= "a" and letter <= "z":
        countlist[ord(letter) - ord("a")][0] += 1

countlist.sort(reverse=True, key=count_letter)

for count in countlist:
    if count[0] > 0:
        print(f"{count[1]}: {count[0]}")
```

## 9. Dictionaries: Word Frequency Counter

A dictionary maps keys to values. Here: word -> count.

**Exercise:** Run the cell. Then change the text and see the new frequencies.

```python
text = "The major advantage of mAID/AID2 here is that you can keep the same genetically engineered line and change when the protein is depleted. That helps separate an early developmental defect with lasting consequences from a requirement for the protein in already specified neurons. One important caveat: developmental age in days in vitro is not equivalent to fetal age, and cerebellar organoids can vary in their differentiation trajectories. Establishing the relevant cell states before selecting induction windows is therefore essential." 

def clean(s):
    news = ""
    s = s.lower()
    for c in s:
        if c >= "a" and c <= "z":
            news += c
        else:
            news += " "
    return news

worddict = {}
for word in clean(text).split():
    worddict[word] = worddict.get(word, 0) + 1

keylist = sorted(worddict.keys())
for key in keylist:
    print(f"{key}: {worddict[key]}")
```

## 10. Dictionaries: Movie Ratings

A dictionary can store lists as values. Here: movie title -> list of ratings.

**Exercise:** Run the cell. Then add your own movie and ratings.

```python
movies = {
    "Monty Python and the Holy Grail": [9, 10, 9.5, 8.5, 3, 7.5, 8],
    "Monty Python's Life of Brian": [10, 10, 0, 9, 1, 8, 7.5, 8, 6, 9],
    "Monty Python's Meaning of Life": [7, 6, 5],
    "And Now For Something Completely Different": [6, 5, 6, 6]
}

keylist = sorted(movies.keys())
for key in keylist:
    avg = sum(movies[key]) / len(movies[key])
    print(f"{key}: {round(avg, 1)}")
```
